```python
dept_margin = (
    customerShopping_df.groupby('Department')[['Revenue', 'Profit']]
    .sum()
    .reset_index()
)

dept_margin['Profit_Margin_%'] = (dept_margin['Profit'] / dept_margin['Revenue']) * 100

dept_margin = dept_margin.sort_values(by='Profit_Margin_%', ascending=False)

print(dept_margin.to_string(index=False))

plt.figure(figsize=(8, 4))
sns.set_theme(style="whitegrid")

ax = sns.barplot(
    data=dept_margin,
    x='Profit_Margin_%',
    y='Department',
    hue='Department',
    legend=False,
)

plt.title('Profit Margin (%) by Department', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Profit Margin (%)', fontsize=12, fontweight='bold')
plt.ylabel('Department', fontsize=12, fontweight='bold')
plt.xlim(0, dept_margin['Profit_Margin_%'].max() * 1.18)

for p in ax.patches:
    width = p.get_width()
    ax.annotate(
        f'{width:.2f}%',
        (width, p.get_y() + p.get_height() / 2.),
        ha='left', va='center',
        fontsize=10, fontweight='bold', color='#333333'
    )

plt.tight_layout()
plt.show()
```

**Time Series Analysis**

```python
customerShopping_df['Purchase_Date'] = pd.to_datetime(customerShopping_df['Purchase_Date'])

monthly_df = (
    customerShopping_df.set_index('Purchase_Date')
    .resample('MS')[['Revenue', 'Profit']]
    .sum()
    .reset_index()
)

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_df['Purchase_Date'],
    monthly_df['Revenue'],
    color='#1f77b4',
    label='Total Revenue',
    linewidth=2.5,
    marker='o'
)
plt.fill_between(
    monthly_df['Purchase_Date'],
    monthly_df['Revenue'],
    color='#1f77b4',
    alpha=0.15
)

plt.plot(
    monthly_df['Purchase_Date'],
    monthly_df['Profit'],
    color='#2ca02c',
    label='Total Profit',
    linewidth=2.5,
    marker='s'
)
plt.fill_between(
    monthly_df['Purchase_Date'],
    monthly_df['Profit'],
    color='#2ca02c',
    alpha=0.25
)

plt.title('Month Start Revenue & Profit Time Trends', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Purchase Date', fontsize=12, fontweight='bold')
plt.ylabel('Amount', fontsize=12, fontweight='bold')
plt.xticks(rotation=30)
plt.legend(loc='upper left', frameon=True, facecolor='white', framealpha=0.9)
plt.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

**Outlier Detection**

```python
numeric_columns = [
    'Quantity',
    'Unit_Price',
    'Revenue',
    'Expense',
    'Profit',
    'Customer_Rating',
    'Employee_Performance_Score'
]
```

```python
customerShopping_df[numeric_columns].describe()
```

```python
plt.figure(figsize=(6, 3))

sns.boxplot(
    data=customerShopping_df[numeric_columns]
)

plt.xticks(rotation=45)
plt.title("Outlier Detection Using Boxplot")
plt.show()
```

```python
numeric_columns = [
    'Quantity',
    'Unit_Price',
    'Revenue',
    'Expense',
    'Profit'
]

for column in numeric_columns:

    Q1 = customerShopping_df[column].quantile(0.25)
    Q3 = customerShopping_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = customerShopping_df[
        (customerShopping_df[column] < lower_bound) |
        (customerShopping_df[column] > upper_bound)
    ]

    print(column)
    print("Lower:", lower_bound)
    print("Upper:", upper_bound)
    print("Outlier count:", len(outliers))
    print()
```
